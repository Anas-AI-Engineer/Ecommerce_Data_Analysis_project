```python
import numpy as np
import pandas as pd

df = pd.read_csv('ecommerce_sales_intermediate.csv')
print(df)

#1Extract Array numpy array 
quantity = df["Quantity"].to_numpy() # varaible banaya df["Quantity"] = csv file sy quantity columns dhondy ga 
                                    #.to_numpy = usey numpy array bany ga 
price = df["Unit_Price_PKR"].to_numpy()# varaible banaya df["Unit_Price_PKR"] = csv file sy quantity columns dhondy ga 
                                       #.to_numpy = usey numpy array bany ga
discount = df["Discount_Percent"].to_numpy()# varaible banaya df["Discount_Percent"] = csv file sy quantity columns dhondy ga
                                            #.to_numpy = usey numpy array bany ga
product = df["Product"].to_numpy()# varaible banaya df["Product"] = csv file sy quantity columns dhondy ga 
                                    #.to_numpy = usey numpy array bany ga
category = df["Category"].to_numpy()# varaible banaya df["Category"] = csv file sy quantity columns dhondy ga 
                                    #.to_numpy = usey numpy array bany ga
city = df["City"].to_numpy()# varaible banaya df["City"] = csv file sy quantity columns dhondy ga 
                                    #.to_numpy = usey numpy array bany ga
payment = df["Payment_Method"].to_numpy()# varaible banaya df["Payment_Method"] = csv file sy quantity columns dhondy ga 
                                    #.to_numpy = usey numpy array bany ga
customer = df["Customer_ID"].to_numpy()# varaible banaya df["Customer_ID"] = csv file sy quantity columns dhondy ga 
                                    #.to_numpy = usey numpy array bany ga
returned = df["Returned"].to_numpy()# varaible banaya df["Returned"] = csv file sy quantity columns dhondy ga 
                                    #.to_numpy = usey numpy array bany ga
#2revenue after discount
print("=" * 40)
revenue = quantity * price * (1 - discount / 100) # varaible banaya aur kaha quantity ko * karo price sy * rows sy - discount / karo 100 sy

print("Total Revenue:", np.sum(revenue)) # total revenue maloom kia np.sum = total maloom krny key leye recall varaible
print("Average Revenue:", np.mean(revenue)) # average revenue maloom kia np.mean = average maloom krny key leye recall varaible
print("Medain Revenue:", np.median(revenue)) # medain revenue maloom kia np.medain = medain maloom krny key leye recall varaible
print("Highest Revenue:", np.argmax(revenue)) # highest revenue maloom kia np.argmax = total maloom krny key leye recall varaible
print("Lowest Revenue:", np.argmin(revenue)) # lowest revenue maloom kia np.argmin = lowest maloom krny key leye recall varaible
print("Total Unit Sold:", np. sum(revenue))  # toatl unit revenue maloom kia np.sum = total maloom krny key leye recall varaible
print("=" * 40)

#3 top 5 orders revenue

top_orders = np.argsort(revenue)[-5:][::-1] # varaible banaya np.argsort = array ki value ko ascending order ma lekhy ga aur kaha (revenue) [-5::]
                                            #= akhri 5 rows ko select karo [::-1] aur unko reverse karo
print("Top 5 Orders:")
for i in top_orders: # for loop lagayi recall karwaya variable ko
    print(product[i], "Revenue:", round(revenue[i],2)) # revenue ka product bhei batana i ko recall krwaya aur kaha agar
                                                       # . ky badd value zada ko tu round 2 krdena matlab serf 2 values dekhana

#4 Category wise performance
print("="*40)
print("Category Analysis:")
for c in np.unique(category): # for loop lagayi c = category ma sy serf = np.unique = serf unique values lena
    mask = category == c, #  mask =true ya false = category
    print(c,
          "|Revenue:", round(np.sum(revenue[mask]),2), # revenue ko sum krwaya aur kah agar . ky baad values zada ho tu serf 2 no dekhana 
          "|Unit Sold:", np.sum(quantity[mask]) # quantity ko sum krraya
         )

# 5city wise analysis
print("="*40)
print("City Analysis:")
for c in np.unique(city): # foor loop lagayi np.unique = serf unique city ko select kry ga varaible ko recall krwaya 
    print(c, ":", round(np.sum(revenue[city == c]),2)) # np.sum = sum krwaya  revenue ko city ko aur round krway agar 
                                                       #. key baad value zada ho tu bs 2 dekhna 

#6 best selling product by quantity
print("="*40)
print("Best Quantity product:")
for p in np.unique(product):# foor loop lagayi np.unique = serf unique product ko select kry ga varaible ko recall krwaya                                         
    print(p, ":", np.sum(quantity[product == p]))# np.sum = sum krwaya  revenue ko city ko aur round krwaya agar
                                                 #. key baad value zada ho tu bs 2 dekhna

#7 Customers Analysis
print("="*40)
unique_customers, counts = np.unique(customer, return_counts=True) #varaible banaya np.unique=serf unique customers bataye ga customers
                                                                   # ko recall krwaye ga return_counts = hr unique value ki counting bhei return karo 
top_customers = np.argmax(counts) # varaible banaya np.argmax=top sy customers ko count karey ga

print("Most unique Customers:", len(unique_customers)) # len = unique customers bataye ga ketny ha
print("Most frequent customers:", unique_customers,(top_customers)) # unique customers ma sy top customers ko dhoondhy ga
print("Orders by this customers:", top_customers,(counts))# customers counts kary ga

print("="*40)
#8 Payment orders analysis
for p in np.unique(payment): # for loop lagayi np.unique = serf unique peyment select karey ga
    print(
        p, "Orders:", np.sum(payment==p) # p ko recall krwaya np.sum = payment ko aur p ko 
    )

#9 Return Rate
print("="*40)
return_mask = returned == "Yes" #returned → saare orders ka return status.
                                #=="Yes" → check karta hai kaunse orders return hue hain.
                                #return_mask → True aur False ki array banata hai

print("Returned Order:", np.sum(return_mask))# return mask ko sum krwaya
print("Return Rate %:", round(np.mean(return_mask)* 100 ,2)) # return_mask ma sy average maloom karo aur * karo 100 sy aur agar value . ky baad value 
                                                             #zada ho tu serf 2 value dekhana 

#10 discount analys
print("="*40)
print("Average Discount:", round(np.mean(discount), 2), "%") # np.mean discount ma sy average maloom karo aur agar value . ky baad value 
                                                             #zada ho tu serf 2 value dekhana 
print("Highest Discount:", np.max(discount), "%")            # np.max = maximium discount maloom kia 

print("="*40)
#11 outlier dectection using iqr
q1 = np.percentile(revenue, 25) # varaible banaya np.percentile = bataye ga revenue ma 25% sy zada ketny revenue ha
q3 = np.percentile(revenue, 75) # varaible banaya np.percentile = bataye ga revenue ma 75% sy zada ketny revenue ha
iqr = q3 - q1                   #iqr= data ky beech ka yahi 25 nd 75 ky darmiyan ka 50% hissa calculate krta ha aur kaha q3 ma sy - krdu q1 ko

outlier = revenue[              # varaible banaya revenue ko recall krwaya 
    (revenue < q1 - 1.5 * iqr) |# kaha revenue agar kam ho q1 sy - krna 1.5 aur * krna iqr sy
    (revenue > q3 + 1.5 * iqr)  # kaha revenue agar zada ho q3 sy + krna 1.5 aur * krna iqr sy
    ]

print("Revene Outlier:",len(outlier)) # outlier key no batey ga
print("Outlier Value:", np.round(outlier, 2)) #recall krwaya outlier ko

print("="*40)
#12 revenue contribution by share
print("Category Revenue Share:")
for c in np.unique(category): # foor loop lagayi aur kaha np.unique=serf unique category batana
    category_revenue = np.sum(revenue[category == c]) # recall krwaya variable np.sum = revenue ko category ko sum krwaya
    share = category_revenue / np.sum(revenue) * 100 # variable banaya aur kaha 
    print(c, ":", round(share, 2), "%")

print("=====ECOMMERCE DATA ANALYSIS COMPLETED=====")
    

    
```
